```python
#3.check whether the given num is prime or not
n=int(input())
for i in range(2,n):
 if(n%i==0):
  print("it is not a prime")
  break
else:
  print("yes it is prime")
```

```python
#4.print prime numbers between given 2 no's
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=" ")
```

```python
#5.print the sum of the digits of a given number
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```
